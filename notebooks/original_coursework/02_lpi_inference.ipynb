```python
# Import necessary libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
from scipy.stats import pearsonr
from sklearn.preprocessing import StandardScaler
import statsmodels.api as sm
from scipy import stats
```

```python
# Load the Excel file
file_path = 'International_LPI_from_2007_to_2023_Cleaned.xlsx'
```

```python
# Load all sheets into a dictionary of DataFrames

sheets_dict = pd.read_excel(file_path, sheet_name=None)
```

```python
# Combine all sheets into one DataFrame
df = pd.concat(sheets_dict.values(), ignore_index=True)
```

## Hypothesis Testing for Means - Mexico LPI score vs Worldwide 2023

## Null (H₀): World mean ≤ Mexico’s LPI

## Alternative (H₁): World mean > Mexico’s LPI

```python
# Hypothesis Test for Means Mexico LPI Score 2023 vs. Global Mean LPI Score 2023 at 1% significance Z score

# From DataFrame isolate 2023 data
world_lpi_2023 = df[df['Year'] == 2023]['LPI Score']

# Parameters
mexico_lpi = df.loc[df['Country'] == 'Mexico', 'LPI Score'].values[0]  # Mexico's LPI Score
sample_mean = world_lpi_2023.mean() # Global LPI Score Mean
population_std_dev = world_lpi_2023.std(ddof=0)  # population std dev
sample_size = len(world_lpi_2023) # Number of countries

# Calculate the z-score
z_score = (sample_mean - mexico_lpi) / (population_std_dev / np.sqrt(sample_size))

# Right tailed One-tailed test
p_value = 1 - stats.norm.cdf(z_score)

print(f"Z-score: {z_score:.3f}")
print(f"P-value: {p_value:.4f}")

# Hypothesis test at alpha = 0.01
alpha = 0.01
if p_value < alpha:
    print(f"Reject the null hypothesis. There is enough evidence to conclude that the worldwide mean LPI ({sample_mean:.2f}) is greater than Mexico's LPI ({mexico_lpi}).")
else:
    print(f"Fail to reject the null hypothesis. There is not enough evidence to conclude that the worldwide mean LPI ({sample_mean:.2f}) is greater than Mexico's LPI ({mexico_lpi}).")
```

```python
# Hypothesis Test for Means Mexico LPI score 2023 vs. Global Mean LPI Score 2023 at 1% significance using T-test

# Filter for 2023
world_lpi_2023 = df.loc[df['Year'] == 2023, 'LPI Score'].astype(float)

# Extract Mexico's 2023 LPI as a float (not a Series)
mexico_lpi_2023 = float(df.loc[(df['Country'] == 'Mexico') & (df['Year'] == 2023), 'LPI Score'].values[0])

# One-sample t-test: compare worldwide sample to Mexico's LPI
t_stat, p_value_two_tailed = stats.ttest_1samp(world_lpi_2023, mexico_lpi_2023)

# Ensure t_stat is a float
t_stat = float(t_stat)

# Convert to one-tailed p-value
if t_stat > 0:
    p_value_one_tailed = p_value_two_tailed / 2
else:
    p_value_one_tailed = 1 - (p_value_two_tailed / 2)

# Print results
print(f"T-statistic: {t_stat:.3f}")
print(f"One-tailed p-value: {p_value_one_tailed:.4f}")

# Significance level
alpha = 0.01

if p_value_one_tailed < alpha:
    print(f"Reject the null hypothesis. There is enough evidence to conclude that the worldwide mean LPI ({world_lpi_2023.mean():.2f}) is greater than Mexico's LPI ({mexico_lpi_2023}).")
else:
    print(f"Fail to reject the null hypothesis. There is not enough evidence to conclude that the worldwide mean LPI ({world_lpi_2023.mean():.2f}) is greater than Mexico's LPI ({mexico_lpi_2023}).")
```
